# Sales Analysis

```python
Category_sales=df.groupby("Category")["Sales"].sum()
print(f"The category has the highest sales is {Category_sales.idxmax()}")
```

```python
sub_category_sales=df.groupby("Sub-Category")["Sales"].sum()
```

```python
print(f"The sub-Category has the highest sales is {sub_category_sales.idxmax()}")
```

```python
states_sales=df.groupby("State")["Sales"].sum()
print(f"The states has highest sales is {states_sales.idxmax()}") 
```

```python
region_sales=df.groupby("Region")["Sales"].sum()
```

```python
print(f"The region has the sales is {region_sales.idxmax()}")
```

```python
# Monthly sales trends 
df["Order_date"]=pd.to_datetime(df["Order Date"])
df["Month"]=df["Order_date"].dt.month
monthly_sales=df.groupby("Month")["Sales"].sum()
plt.plot(monthly_sales,color="red",alpha=1)
plt.title("Monthly Sales trends")
plt.xlabel("Month")
plt.ylabel("Sales")
```

```python
product_sales=df.groupby("Product Name")["Sales"].sum()
print(f"The product with highest sales is {product_sales.idxmax()}")
```

# Profit Analysis

```python
category_profit=df.groupby("Category")["Profit"].sum()
print(f"The category has the highest profit is { category_profit.idxmax()}")
```

```python
subcategory_profit=df.groupby("Sub-Category")["Profit"].sum()
print(f"The sub category has the highest profit is { subcategory_profit.idxmax()}")
```

```python
product_profit=df.groupby("Product Name")["Profit"].sum()
print(f"The product has the highest profit is {product_profit.idxmax()}")
```

```python
print ( f"The product have negative product name is {df["Product Name"][df["Profit"].idxmin()]}")
```

```python
# Coorelation between Sales and profit columns
print(f"The correlation between sales and Profit is  {df["Sales"].corr(df["Profit"])}")
# Now draw the scatterplot for better visualization
sns.scatterplot(data=df,x="Sales",y="Profit",hue="Profit")
plt.title("Scatter plot between Sales and Profit columns")
plt.show()
```

# Customer Analysis

```python
# The top 10 customers by sales
customer_sales=df.groupby("Customer Name")["Sales"].sum()
print("The top 10 customers by sales :")

print(customer_sales.sort_values(ascending=False).head(10))
```

```python
# Top 10 customers By profit
customer_profit=df.groupby("Customer Name")["Profit"].sum()
print("Top 10 customers by profit")
print(customer_profit.sort_values(ascending=False).head(10))
```

```python
# Which customer segment generates the most sales
segment_sales=df.groupby("Segment")["Sales"].sum()
print(f"The customer segment which generates the most sales  is  { segment_sales.idxmax()}")
```

```python
# Which customer segment generates the most profit
segment_profit=df.groupby("Segment")["Profit"].sum()
print(f"The customer segment which generates the most profit is  { segment_profit.idxmax()}")
```

# Shipping Analysis

```python
# Which shipping mode is used most frequently
print(f"Most frequently used shipping mode is { df["Ship Mode"].max()}")
```

```python
# Which shipping mode generates the highets sales
shipping_sales=df.groupby("Ship Mode")["Sales"].sum()
print(f"The shipping mode has the highest sales { shipping_sales.idxmax()}")
```

```python
# Does shipping time effect profit
df["Ship_Date"]=pd.to_datetime(df["Ship Date"])
df["Shipping Time"]=(df["Order_date"]-df["Ship_Date"]).dt.days
# Find correlation
print(df["Shipping Time"].corr(df["Profit"]))
# Now plot the scatter plot 
sns.scatterplot(data=df,x=df["Shipping Time"],y=df["Profit"])
plt.title("Shipping Time v/s Profit")
```
